### 1. What is Data Ingestion?

*   **Definition**: The process of obtaining, importing, and loading data into a system for immediate use or storage in a database for later use.
*   **Sources**:
    *   CSV / Excel files (local or remote)
    *   Databases (SQL, NoSQL)
    *   Web APIs (JSON/XML responses)
    *   Web Scraping
    *   Built-in datasets (e.g., `sklearn`, `seaborn`)
*   **Key Tool**: `pandas` library — provides `read_csv()`, `read_excel()`, `read_json()`, `read_sql()`, and more.

### 2. What is Data Cleaning?

*   **Definition**: The process of detecting and correcting (or removing) inaccurate, incomplete, or irrelevant records from a dataset.
*   **Why it matters**: Raw data is often messy; data quality directly impacts the reliability of analysis and models.
*   **Common Issues**:
    *   Missing values (`NaN`)
    *   Duplicate rows
    *   Incorrect data types
    *   Inconsistent formatting (e.g., `' male'` vs `'Male'`)
    *   Outliers

### 3. Handling Missing Values

*   **Detection**: `df.isnull()`, `df.isnull().sum()`
*   **Strategies**:
    *   **Drop**: `df.dropna()` — removes rows/columns with missing values. Safe when missing data is minimal.
    *   **Fill with constant**: `df.fillna(0)` — replaces NaN with a fixed value.
    *   **Fill with mean/median/mode**: Commonly used for numerical columns to preserve distribution.
    *   **Forward/backward fill**: `df.ffill()` / `df.bfill()` — useful for time-series data.

### 4. Handling Duplicate Data

*   **Detection**: `df.duplicated()` — returns a Boolean Series marking duplicate rows.
*   **Removal**: `df.drop_duplicates()` — removes duplicate rows, keeping the first occurrence by default.
*   **Why it matters**: Duplicates can skew statistics and model training, leading to biased results.

### 5. Data Type Conversion and Standardization

*   **Checking types**: `df.dtypes` — displays the data type of each column.
*   **Conversion**: `df['col'].astype(int)`, `pd.to_datetime()`, `pd.to_numeric()` — convert columns to appropriate types.
*   **String Standardization**:
    *   `.str.strip()` — removes leading/trailing whitespace.
    *   `.str.lower()` / `.str.upper()` — normalizes case.
    *   `.str.replace()` — replaces specific characters or patterns.
*   **Renaming Columns**: `df.rename(columns={'old': 'new'})` — for clarity and consistency.

### Demonstrations of Data Ingestion and Cleaning

#### 1. Data Ingestion — Creating a Dataset (Simulating CSV Ingestion)

```python
import pandas as pd
import numpy as np
import io

# Simulating a raw CSV data string (as if read from a file)
raw_csv = """Name,Age,Gender,Salary,Department
Alice,28, Female,55000,HR
Bob,35,Male,72000,IT
Charlie,,Male,48000,Finance
Alice,28, Female,55000,HR
David,45,Male,,IT
Eva,29,female,61000,Marketing
Frank,38,Male,83000,IT
Grace,,Female,47000,Finance
Heidi,31, FEMALE,52000,HR
Ivan,27,Male,44000,
"""

# Ingesting data using pandas read_csv
df = pd.read_csv(io.StringIO(raw_csv))

print("--- Raw Ingested Data ---")
print(df)
print(f"\nShape: {df.shape} (rows, columns)")
```

#### 2. Initial Data Inspection

```python
print("--- Basic Info ---")
df.info()

print("\n--- First 5 Rows ---")
print(df.head())

print("\n--- Statistical Summary ---")
print(df.describe())

print("\n--- Column Data Types ---")
print(df.dtypes)
```

#### 3. Detecting Missing Values

```python
print("--- Missing Values Per Column ---")
print(df.isnull().sum())

print("\n--- Missing Value Percentage ---")
missing_pct = (df.isnull().sum() / len(df)) * 100
print(missing_pct)

print("\n--- Rows with Missing Values ---")
print(df[df.isnull().any(axis=1)])
```

#### 4. Handling Missing Values

```python
df_clean = df.copy()

# Fill missing 'Age' with the median age (robust to outliers)
median_age = df_clean['Age'].median()
df_clean['Age'] = df_clean['Age'].fillna(median_age)
print(f"Filled missing Age values with median: {median_age}")

# Fill missing 'Salary' with the mean salary
mean_salary = df_clean['Salary'].mean()
df_clean['Salary'] = df_clean['Salary'].fillna(round(mean_salary, 2))
print(f"Filled missing Salary values with mean: {round(mean_salary, 2)}")

# Fill missing 'Department' with 'Unknown'
df_clean['Department'] = df_clean['Department'].fillna('Unknown')
print("Filled missing Department values with 'Unknown'")

print("\n--- Missing Values After Filling ---")
print(df_clean.isnull().sum())
```

#### 5. Detecting and Removing Duplicate Rows

```python
print("--- Duplicate Rows ---")
print(df_clean[df_clean.duplicated()])

print(f"\nNumber of duplicate rows: {df_clean.duplicated().sum()}")

# Remove duplicates
df_clean = df_clean.drop_duplicates()
df_clean = df_clean.reset_index(drop=True)

print(f"\nShape after removing duplicates: {df_clean.shape}")
print("\n--- Data After Removing Duplicates ---")
print(df_clean)
```

#### 6. Standardizing String/Categorical Columns

```python
print("--- Gender values before cleaning ---")
print(df_clean['Gender'].unique())

# Strip whitespace and convert to title case for consistency
df_clean['Gender'] = df_clean['Gender'].str.strip().str.title()
df_clean['Name'] = df_clean['Name'].str.strip().str.title()
df_clean['Department'] = df_clean['Department'].str.strip().str.title()

print("\n--- Gender values after cleaning ---")
print(df_clean['Gender'].unique())

print("\n--- Data After String Standardization ---")
print(df_clean)
```

#### 7. Correcting Data Types

```python
print("--- Data Types Before Conversion ---")
print(df_clean.dtypes)

# Convert Age to integer and Salary to integer
df_clean['Age'] = df_clean['Age'].astype(int)
df_clean['Salary'] = df_clean['Salary'].astype(int)

print("\n--- Data Types After Conversion ---")
print(df_clean.dtypes)

print("\n--- Final Cleaned Dataset ---")
print(df_clean)
```

#### 8. Summary Statistics on Cleaned Data

```python
print("--- Final Statistical Summary ---")
print(df_clean.describe())

print("\n--- Department-wise Average Salary ---")
print(df_clean.groupby('Department')['Salary'].mean().round(2))

print("\n--- Gender Distribution ---")
print(df_clean['Gender'].value_counts())

print("\n--- Data is now clean and ready for analysis! ---")
```
