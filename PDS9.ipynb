### 1. What is a Data Science Pipeline? Explain its Key Stages with Examples.

* **Definition**:
    * A Data Science Pipeline is a **structured, end-to-end workflow** that transforms raw data into actionable insights or a deployable machine learning model.
    * It ensures reproducibility, modularity, and systematic processing of data.

* **Key Stages**:

    1. **Data Collection**:
        * Gathering raw data from sources such as databases, APIs, CSV files, web scraping, or sensors.
        * Example: Loading a student dataset from a CSV file using `pandas.read_csv()`.

    2. **Data Cleaning / Preprocessing**:
        * Handling missing values, removing duplicates, correcting data types, and fixing inconsistencies.
        * Example: Filling missing age values with the column mean using `df['Age'].fillna(df['Age'].mean())`.

    3. **Exploratory Data Analysis (EDA)**:
        * Visualizing and summarizing data to discover patterns, distributions, and relationships.
        * Example: Plotting a histogram of exam scores to understand their distribution.

    4. **Feature Engineering**:
        * Creating new features, encoding categorical variables, and transforming existing features to improve model performance.
        * Example: Creating a `Score_per_Hour` feature from `Exam_Score / Study_Hours`.

    5. **Feature Scaling**:
        * Normalizing or standardizing numeric features so that all variables are on a comparable scale.
        * Example: Applying `StandardScaler` to bring all features to zero mean and unit variance.

    6. **Model Building**:
        * Selecting and training a machine learning algorithm on the processed data.
        * Example: Training a Linear Regression model using `sklearn.linear_model.LinearRegression`.

    7. **Model Evaluation**:
        * Measuring the model's performance using appropriate metrics.
        * Example: Computing R2, RMSE, and MAE to evaluate regression quality.

    8. **Deployment / Communication**:
        * Deploying the model as an API or dashboard, or communicating findings to stakeholders.
        * Example: Exporting model predictions to a report or hosting via Flask/FastAPI.

* **Why Pipelines Matter**:
    * Prevent data leakage between training and testing.
    * Make workflows reproducible and easy to maintain.
    * Allow end-to-end automation of the ML process.

### 2. What is Exploratory Data Analysis (EDA)? How do Univariate and Bivariate Analyses Help?

* **Exploratory Data Analysis (EDA)**:
    * An approach to **summarizing, visualizing, and understanding** the main characteristics of a dataset before formal modeling.
    * Introduced by statistician John Tukey.
    * Helps detect outliers, understand distributions, identify patterns, and reveal relationships between variables.
    * Tools: `describe()`, `info()`, histograms, box plots, scatter plots, heatmaps.

* **Univariate Analysis** (one variable at a time):
    * Focuses on the **distribution and summary** of a single variable.
    * Answers: What is the range? Is it skewed? Are there outliers? What is the central tendency?
    * Techniques:
        * For numerical: histogram, box plot, KDE plot, mean/median/mode/std.
        * For categorical: bar chart, frequency table, pie chart.
    * Example: Plotting a histogram of student ages to check for unusual values.

* **Bivariate Analysis** (two variables together):
    * Examines the **relationship between two variables**.
    * Answers: Do variables move together? Is there a trend? Are groups different?
    * Techniques:
        * Numerical vs Numerical: scatter plot, Pearson correlation.
        * Numerical vs Categorical: box plot, violin plot, group means.
        * Categorical vs Categorical: cross-tabulation, grouped bar chart.
    * Example: Scatter plot of Study Hours vs Exam Score to check linear relationship.

* **Combined Benefit**:
    * Univariate analysis cleans and understands individual features.
    * Bivariate analysis identifies which features are predictive of the target variable.
    * Together they guide feature selection, transformation decisions, and model choice.

### 3. Differentiate Between Label Encoding and OneHot Encoding. When Should Each Be Used?

* **Label Encoding**:
    * Converts each category into a unique **integer** value.
    * Example: `['Low', 'Medium', 'High']` → `[0, 1, 2]`.
    * Preserves ordinal (ranked) information implicitly.
    * Introduces **artificial ordering** — the model may assume High (2) > Medium (1) > Low (0).
    * Best suited for **ordinal variables** (naturally ordered categories) and tree-based models.

* **One-Hot Encoding (OHE)**:
    * Creates a **separate binary column** (0 or 1) for each unique category.
    * Example: `['Red', 'Green', 'Blue']` → three columns: `Is_Red`, `Is_Green`, `Is_Blue`.
    * No artificial ordering is introduced.
    * Best suited for **nominal variables** (no natural order) used in distance-based or linear models.

| Feature            | Label Encoding               | One-Hot Encoding                  |
|--------------------|------------------------------|-----------------------------------|
| Output             | Single integer column        | Multiple binary columns           |
| Ordering Implied   | Yes (can be misleading)      | No                                |
| Dimensionality     | No increase                  | Increases with number of classes  |
| Best For           | Ordinal data, tree models    | Nominal data, linear/ML models    |
| Example            | Grade: Low=0, Med=1, High=2  | Color: Red=[1,0,0], Green=[0,1,0] |

* **When to use which**:
    * Use **Label Encoding** for: Education level (No School < Undergrad < Postgrad), Risk level (Low/Med/High).
    * Use **One-Hot Encoding** for: City names, Blood type, Gender, Product category.

### 4. Why is Feature Scaling Required? Compare Normalization and Standardization.

* **Why Feature Scaling is Required**:
    * Machine learning algorithms that rely on distance calculations (KNN, SVM, K-Means) or gradient descent (Linear/Logistic Regression, Neural Networks) are **sensitive to the scale of features**.
    * Without scaling, a feature with a large range (e.g., income: 10,000–1,000,000) will dominate a feature with a small range (e.g., age: 18–60).
    * Scaling ensures all features contribute **equally** to model training.
    * Tree-based models (Decision Trees, Random Forests) are generally scale-invariant and do NOT require scaling.

* **Normalization (Min-Max Scaling)**:
    * Rescales data to a **fixed range**, typically [0, 1].
    * Formula: `X_scaled = (X - X_min) / (X_max - X_min)`
    * Sensitive to outliers — one extreme value compresses all others.
    * Best when data is not normally distributed and bounded range is needed.
    * Used in: Neural Networks, image pixel normalization, KNN.

* **Standardization (Z-score Scaling)**:
    * Transforms data to have **mean = 0** and **standard deviation = 1**.
    * Formula: `X_scaled = (X - mean) / std`
    * More robust to outliers than normalization.
    * Data retains its distribution shape (does not bound to [0,1]).
    * Used in: Linear Regression, Logistic Regression, SVM, PCA.

| Feature           | Normalization (Min-Max)     | Standardization (Z-score)     |
|-------------------|-----------------------------|-------------------------------|
| Output Range      | [0, 1]                      | Unbounded (mean=0, std=1)     |
| Formula           | (X - min) / (max - min)     | (X - mean) / std              |
| Outlier Sensitive | Yes                         | Less sensitive                |
| Assumes Normality | No                          | Preferred for normal data     |
| Use Case          | Neural nets, KNN, images    | Linear models, SVM, PCA       |

### 5. What is Feature Engineering? How Does Creating New Features Improve Model Performance?

* **Definition**:
    * Feature Engineering is the process of using **domain knowledge and data transformations** to create, modify, or select features that make machine learning algorithms work better.
    * It bridges the gap between raw data and a model that can learn effectively.
    * Often described as the most impactful step in the ML pipeline.

* **Common Feature Engineering Techniques**:

    1. **Creating Interaction Features**:
        * Combining two features to capture relationships not visible individually.
        * Example: `BMI = Weight / Height^2` from separate weight and height columns.

    2. **Polynomial Features**:
        * Adding squared or cubed versions of features to capture non-linear patterns.
        * Example: `Area = Length * Width`, `Length^2`.

    3. **Date/Time Decomposition**:
        * Extracting year, month, day, hour, weekday from a timestamp.
        * Example: Splitting `2024-01-15` into `Year=2024`, `Month=1`, `Day=15`, `Is_Weekend=False`.

    4. **Binning / Discretization**:
        * Converting continuous variables into categorical bins.
        * Example: Age → `Young (0-25)`, `Middle (26-50)`, `Senior (51+)`.

    5. **Aggregation Features**:
        * Computing group-wise statistics to add context.
        * Example: `Avg_Score_per_Department` in an employee dataset.

    6. **Log/Square Root Transformation**:
        * Reducing skewness in highly skewed features.
        * Example: `log(Income)` to reduce right skew.

* **How New Features Improve Performance**:
    * Help linear models capture **non-linear patterns** without changing the algorithm.
    * Reduce **noise** by combining related weak signals into one strong feature.
    * **Encode domain knowledge** that raw data alone cannot express.
    * Improve **interpretability** — a `Debt_to_Income_Ratio` feature is more meaningful than raw income alone.

* **Example**:
    * In a student performance dataset, raw features: `Study_Hours`, `Exam_Score`.
    * Engineered feature: `Score_per_Hour = Exam_Score / Study_Hours` — captures study efficiency.
    * This new feature may correlate more strongly with future performance than either raw feature alone.

### Complete Data Science Pipeline Demonstration

#### Stage 1: Data Collection and Overview

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# ── Simulate a realistic student performance dataset ──
np.random.seed(42)
n = 150

df = pd.DataFrame({
    'Student_ID'   : range(1, n + 1),
    'Age'          : np.random.randint(17, 25, n),
    'Gender'       : np.random.choice(['Male', 'Female'], n),
    'Study_Hours'  : np.round(np.random.uniform(1, 10, n), 1),
    'Sleep_Hours'  : np.round(np.random.uniform(4, 9, n), 1),
    'Attendance_Pct': np.round(np.random.uniform(50, 100, n), 1),
    'Grade'        : np.random.choice(['Low', 'Medium', 'High'], n, p=[0.25, 0.45, 0.30]),
    'Exam_Score'   : np.round(np.random.normal(65, 15, n).clip(20, 100), 1)
})

# Inject missing values intentionally
df.loc[np.random.choice(n, 10, replace=False), 'Sleep_Hours'] = np.nan
df.loc[np.random.choice(n, 8,  replace=False), 'Attendance_Pct'] = np.nan

print("=" * 55)
print("  STAGE 1: DATA COLLECTION AND OVERVIEW")
print("=" * 55)
print(f"\nDataset Shape: {df.shape[0]} rows x {df.shape[1]} columns")
print("\nFirst 5 Rows:")
print(df.head())
print("\nData Types and Non-Null Counts:")
print(df.info())
print("\nBasic Statistics:")
print(df.describe().round(2))
```

#### Stage 2: Data Cleaning — Handling Missing Values and Duplicates

```python
print("=" * 55)
print("  STAGE 2: DATA CLEANING")
print("=" * 55)

# Missing values before cleaning
print("\nMissing Values Before Cleaning:")
missing = df.isnull().sum()
missing_pct = (df.isnull().mean() * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
print(missing_df[missing_df['Missing Count'] > 0])

# Fill missing numerical values with median
df['Sleep_Hours']    = df['Sleep_Hours'].fillna(df['Sleep_Hours'].median())
df['Attendance_Pct'] = df['Attendance_Pct'].fillna(df['Attendance_Pct'].median())

# Check for duplicates
print(f"\nDuplicate Rows: {df.duplicated().sum()}")

# Verify cleaning
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
print(f"\nDataset Shape After Cleaning: {df.shape}")
print("\nCleaned Dataset Sample:")
print(df.head())
```

#### Stage 3: Exploratory Data Analysis — Univariate and Bivariate

```python
print("=" * 55)
print("  STAGE 3: EXPLORATORY DATA ANALYSIS (EDA)")
print("=" * 55)

# ── Univariate Analysis ──
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
fig.suptitle('Univariate Analysis', fontsize=15, fontweight='bold')

num_cols = ['Age', 'Study_Hours', 'Sleep_Hours', 'Attendance_Pct', 'Exam_Score']
colors   = ['steelblue', 'coral', 'mediumseagreen', 'mediumpurple', 'goldenrod']

for ax, col, color in zip(axes.flatten()[:5], num_cols, colors):
    ax.hist(df[col], bins=20, color=color, edgecolor='black', alpha=0.85)
    ax.axvline(df[col].mean(),   color='red',   linestyle='--', linewidth=1.5, label=f'Mean={df[col].mean():.1f}')
    ax.axvline(df[col].median(), color='black', linestyle=':',  linewidth=1.5, label=f'Median={df[col].median():.1f}')
    ax.set_title(col)
    ax.set_xlabel('Value')
    ax.set_ylabel('Frequency')
    ax.legend(fontsize=8)
    ax.grid(linestyle='--', alpha=0.5)

# Categorical: Grade distribution
grade_counts = df['Grade'].value_counts()
axes[1, 2].bar(grade_counts.index, grade_counts.values,
               color=['coral', 'steelblue', 'mediumseagreen'], edgecolor='black', alpha=0.85)
axes[1, 2].set_title('Grade Distribution (Categorical)')
axes[1, 2].set_xlabel('Grade')
axes[1, 2].set_ylabel('Count')
axes[1, 2].grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# ── Bivariate Analysis ──
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('Bivariate Analysis', fontsize=15, fontweight='bold')

# 1. Study Hours vs Exam Score
axes[0].scatter(df['Study_Hours'], df['Exam_Score'],
                color='steelblue', alpha=0.6, edgecolor='black', s=50)
m, b = np.polyfit(df['Study_Hours'], df['Exam_Score'], 1)
x_fit = np.linspace(df['Study_Hours'].min(), df['Study_Hours'].max(), 100)
axes[0].plot(x_fit, m * x_fit + b, color='red', linewidth=2, label='Trend Line')
r_val, _ = stats.pearsonr(df['Study_Hours'], df['Exam_Score'])
axes[0].set_title(f'Study Hours vs Exam Score\nr = {r_val:.3f}')
axes[0].set_xlabel('Study Hours')
axes[0].set_ylabel('Exam Score')
axes[0].legend()
axes[0].grid(linestyle='--', alpha=0.5)

# 2. Gender vs Exam Score (Box Plot)
gender_groups = [df[df['Gender'] == g]['Exam_Score'].values for g in ['Male', 'Female']]
bp = axes[1].boxplot(gender_groups, labels=['Male', 'Female'], patch_artist=True,
                     boxprops=dict(facecolor='lightblue'),
                     medianprops=dict(color='red', linewidth=2))
axes[1].set_title('Gender vs Exam Score')
axes[1].set_xlabel('Gender')
axes[1].set_ylabel('Exam Score')
axes[1].grid(linestyle='--', alpha=0.5)

# 3. Attendance vs Exam Score
axes[2].scatter(df['Attendance_Pct'], df['Exam_Score'],
                color='mediumseagreen', alpha=0.6, edgecolor='black', s=50)
m2, b2 = np.polyfit(df['Attendance_Pct'], df['Exam_Score'], 1)
x_fit2 = np.linspace(df['Attendance_Pct'].min(), df['Attendance_Pct'].max(), 100)
axes[2].plot(x_fit2, m2 * x_fit2 + b2, color='red', linewidth=2, label='Trend Line')
r_val2, _ = stats.pearsonr(df['Attendance_Pct'], df['Exam_Score'])
axes[2].set_title(f'Attendance % vs Exam Score\nr = {r_val2:.3f}')
axes[2].set_xlabel('Attendance %')
axes[2].set_ylabel('Exam Score')
axes[2].legend()
axes[2].grid(linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# Correlation Matrix Heatmap
plt.figure(figsize=(8, 6))
corr = df[num_cols].corr(method='pearson')
sns.heatmap(corr, annot=True, fmt='.3f', cmap='coolwarm',
            center=0, linewidths=0.5, square=True)
plt.title('Pearson Correlation Heatmap', fontsize=13)
plt.tight_layout()
plt.show()
```

#### Stage 4: Feature Engineering — Encoding and New Features

```python
from sklearn.preprocessing import LabelEncoder

print("=" * 55)
print("  STAGE 4: FEATURE ENGINEERING")
print("=" * 55)

df_eng = df.copy()

# ── Label Encoding: Grade (ordinal — Low < Medium < High) ──
grade_order = {'Low': 0, 'Medium': 1, 'High': 2}
df_eng['Grade_Encoded'] = df_eng['Grade'].map(grade_order)
print("\nLabel Encoding — Grade (Ordinal):")
print(df_eng[['Grade', 'Grade_Encoded']].drop_duplicates().sort_values('Grade_Encoded'))

# ── One-Hot Encoding: Gender (nominal — no order) ──
df_eng = pd.get_dummies(df_eng, columns=['Gender'], prefix='Gender', drop_first=False)
print("\nOne-Hot Encoding — Gender (Nominal):")
print(df_eng[['Gender_Male', 'Gender_Female']].head())

# ── Creating New Engineered Features ──
# 1. Score per Study Hour — study efficiency
df_eng['Score_per_Hour'] = (df_eng['Exam_Score'] / df_eng['Study_Hours']).round(2)

# 2. Study-Sleep Balance Ratio
df_eng['Study_Sleep_Ratio'] = (df_eng['Study_Hours'] / df_eng['Sleep_Hours']).round(2)

# 3. Age Group (Binning)
df_eng['Age_Group'] = pd.cut(df_eng['Age'], bins=[16, 19, 22, 25],
                              labels=['Junior', 'Mid', 'Senior'])

# 4. High Attendance Flag
df_eng['High_Attendance'] = (df_eng['Attendance_Pct'] >= 75).astype(int)

print("\nEngineered Features Added:")
new_features = ['Score_per_Hour', 'Study_Sleep_Ratio', 'Age_Group', 'High_Attendance']
print(df_eng[new_features].head(10))

print("\nNew Feature Correlations with Exam_Score:")
for feat in ['Score_per_Hour', 'Study_Sleep_Ratio', 'High_Attendance']:
    r, p = stats.pearsonr(df_eng[feat], df_eng['Exam_Score'])
    print(f"  {feat:25s}: r = {r:.4f}  (p = {p:.4f})")
```

#### Stage 5: Feature Scaling — Normalization and Standardization

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

print("=" * 55)
print("  STAGE 5: FEATURE SCALING")
print("=" * 55)

scale_cols = ['Study_Hours', 'Sleep_Hours', 'Attendance_Pct', 'Age']
X_raw = df_eng[scale_cols].copy()

# Min-Max Normalization
minmax = MinMaxScaler()
X_norm = pd.DataFrame(minmax.fit_transform(X_raw), columns=scale_cols)

# Z-score Standardization
zscore = StandardScaler()
X_std  = pd.DataFrame(zscore.fit_transform(X_raw), columns=scale_cols)

print("\nBefore Scaling (first 5 rows):")
print(X_raw.head())
print("\nAfter Normalization (Min-Max) — range [0, 1]:")
print(X_norm.head().round(4))
print("\nAfter Standardization (Z-score) — mean=0, std=1:")
print(X_std.head().round(4))

print("\nNormalization Range Check:")
print(X_norm.describe().loc[['min', 'max']].round(4))
print("\nStandardization Stats Check:")
print(X_std.describe().loc[['mean', 'std']].round(4))

# Visual Comparison
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
fig.suptitle('Feature Scaling Comparison (Study_Hours)', fontsize=13, fontweight='bold')

for ax, data, title, color in zip(
    axes,
    [X_raw['Study_Hours'], X_norm['Study_Hours'], X_std['Study_Hours']],
    ['Original',           'Normalized [0,1]',    'Standardized (Z-score)'],
    ['steelblue',          'coral',                'mediumseagreen']
):
    ax.hist(data, bins=20, color=color, edgecolor='black', alpha=0.85)
    ax.axvline(data.mean(), color='black', linestyle='--', linewidth=1.5,
               label=f'Mean={data.mean():.2f}')
    ax.set_title(title)
    ax.set_xlabel('Value')
    ax.set_ylabel('Frequency')
    ax.legend(fontsize=9)
    ax.grid(linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

#### Stage 6: Model Building and Evaluation

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

print("=" * 55)
print("  STAGE 6: MODEL BUILDING AND EVALUATION")
print("=" * 55)

# Select features and target
feature_cols = ['Study_Hours', 'Sleep_Hours', 'Attendance_Pct',
                'Grade_Encoded', 'Score_per_Hour', 'High_Attendance']
X = df_eng[feature_cols].copy()
y = df_eng['Exam_Score']

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train-Test Split (80:20)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
print(f"\nTrain set size : {X_train.shape[0]} samples")
print(f"Test  set size : {X_test.shape[0]} samples")

# Train Model
model = LinearRegression()
model.fit(X_train, y_train)

# Predictions
y_pred_train = model.predict(X_train)
y_pred_test  = model.predict(X_test)

# Evaluation
def evaluate(y_true, y_pred, split_name):
    r2   = r2_score(y_true, y_pred)
    mse  = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mae  = mean_absolute_error(y_true, y_pred)
    print(f"\n  {split_name} Results:")
    print(f"    R-squared : {r2:.4f}  → Explains {r2*100:.2f}% of variance")
    print(f"    MSE       : {mse:.4f}")
    print(f"    RMSE      : {rmse:.4f}")
    print(f"    MAE       : {mae:.4f}")

evaluate(y_train, y_pred_train, 'Training Set')
evaluate(y_test,  y_pred_test,  'Test Set    ')

# Feature Importance (Coefficients)
print("\n  Model Coefficients (Feature Importance):")
coef_df = pd.DataFrame({'Feature': feature_cols, 'Coefficient': model.coef_})
coef_df = coef_df.reindex(coef_df['Coefficient'].abs().sort_values(ascending=False).index)
print(coef_df.to_string(index=False))

# Actual vs Predicted Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Model Evaluation', fontsize=13, fontweight='bold')

# Actual vs Predicted
axes[0].scatter(y_test, y_pred_test, color='steelblue', edgecolor='black', alpha=0.7, s=60)
lims = [min(y_test.min(), y_pred_test.min()) - 2,
        max(y_test.max(), y_pred_test.max()) + 2]
axes[0].plot(lims, lims, 'r--', linewidth=2, label='Perfect Prediction')
axes[0].set_title(f'Actual vs Predicted (Test Set)\nR2 = {r2_score(y_test, y_pred_test):.4f}')
axes[0].set_xlabel('Actual Exam Score')
axes[0].set_ylabel('Predicted Exam Score')
axes[0].legend()
axes[0].grid(linestyle='--', alpha=0.5)

# Residuals Plot
residuals = y_test - y_pred_test
axes[1].scatter(y_pred_test, residuals, color='coral', edgecolor='black', alpha=0.7, s=60)
axes[1].axhline(0, color='black', linestyle='--', linewidth=1.5)
axes[1].set_title('Residual Plot (Test Set)')
axes[1].set_xlabel('Predicted Values')
axes[1].set_ylabel('Residuals')
axes[1].grid(linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

#### Stage 7: Pipeline Summary

```python
print("=" * 55)
print("  STAGE 7: COMPLETE PIPELINE SUMMARY")
print("=" * 55)

summary = [
    ("Stage 1", "Data Collection",     f"{df.shape[0]} rows x {df.shape[1]} columns loaded"),
    ("Stage 2", "Data Cleaning",       "Handled 18 missing values (median imputation), 0 duplicates found"),
    ("Stage 3", "EDA",                 "Univariate histograms + bivariate scatter/box plots + heatmap"),
    ("Stage 4", "Feature Engineering", "Label Encoding (Grade), One-Hot (Gender), 4 new features created"),
    ("Stage 5", "Feature Scaling",     "Min-Max Normalization and Z-score Standardization applied"),
    ("Stage 6", "Model + Evaluation",  f"LinearRegression | Test R2 = {r2_score(y_test, y_pred_test):.4f} | RMSE = {np.sqrt(mean_squared_error(y_test, y_pred_test)):.4f}"),
]

print()
for stage, name, detail in summary:
    print(f"  [{stage}] {name}")
    print(f"           {detail}")
    print()

print("Pipeline execution complete.")
print("All stages integrated into a single reproducible workflow.")
```
