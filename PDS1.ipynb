### 1. Python Sequence Data Types, Indexing, and Slicing

*   **Definition**: Ordered collections of items.
*   **Types**:
    *   `str` (String): Immutable sequence of characters.
    *   `list` (List): Mutable sequence of items.
    *   `tuple` (Tuple): Immutable sequence of items.
    *   `range` (Range): Immutable sequence of numbers.
*   **Indexing**: Access individual elements using zero-based integer indices.
    *   Positive: `[0, 1, 2, ...]` from the beginning.
    *   Negative: `[-1, -2, -3, ...]` from the end.
    *   Works identically across strings, lists, tuples, and ranges.
*   **Slicing**: Extract sub-sequences `[start:stop:step]`.
    *   `start`: Inclusive beginning index (default 0).
    *   `stop`: Exclusive ending index (default end).
    *   `step`: Increment (default 1).
    *   Always returns a new sequence of the same type.
    *   Works identically across strings, lists, tuples, and ranges.

### 2. Difference Between List and Tuple & Tuple Preference

*   **Lists**: Mutable, uses `[]`, elements can be changed (added, removed, modified).
*   **Tuples**: Immutable, uses `()`, elements cannot be changed after creation.
*   **Why prefer Tuples?**
    *   **Data Integrity**: Guarantees data remains constant.
    *   **Performance**: Slightly faster than lists.
    *   **Dictionary Keys**: Being immutable, tuples can be used as dictionary keys.
    *   **Function Arguments/Return Values**: Ideal for fixed collections.
    *   **Thread Safety**: Inherently safe in multi-threaded environments.

### 3. Python Arrays vs. Lists & When to use Arrays

*   **Lists**:
    *   Heterogeneous (can store different data types).
    *   Stores references to objects.
    *   Less memory-efficient for large numerical data.
*   **Arrays (`array.array`)**:
    *   Homogeneous (stores a single, specified data type).
    *   Stores values directly in contiguous memory.
    *   More memory-efficient for large numerical data.
*   **When to use `array` module**:
    *   Storing large amounts of numerical data efficiently.
    *   Interacting with C/C++ libraries (expect contiguous data).
    *   Performance-critical numerical operations.
    *   Working with raw binary data.

### 4. How Dictionaries Work & Why Lookups are Faster

*   **Internal Working**: Implemented as **hash tables**.
    *   **Hashing**: Keys are converted into a hash value (integer).
    *   **Bucket Assignment**: Hash value maps to an index (bucket) in an internal array.
    *   **Collision Handling**: Manages multiple key-value pairs mapping to the same bucket.
    *   Each entry typically stores (key, value, hash).
*   **Why Faster Lookups?**
    *   **Direct Access via Hashing**: On average, O(1) (constant time) complexity. The hash leads directly to the data's location.
    *   **Reduced Comparisons**: Fewer comparisons needed compared to linear searches in lists/tuples.
    *   **Space-Time Trade-off**: Uses more memory to achieve speed.

### 5. Difference Between Sets and Lists & Set Utility

*   **Lists**:
    *   Ordered collection.
    *   Mutable.
    *   Allows duplicate elements.
*   **Sets**:
    *   Unordered collection.
    *   Mutable (can add/remove elements), but elements themselves must be immutable.
    *   **Does not allow duplicate elements**; each element is unique.
*   **How Sets Handle Duplicates**: Automatically removes duplicates. Trying to add an existing element has no effect.
*   **Useful Scenarios for Sets**:
    *   **Removing Duplicates**: Efficiently get unique items from a collection.
    *   **Membership Testing**: Very fast `in` operator (O(1) average).
    *   **Mathematical Set Operations**: Union, Intersection, Difference, Symmetric Difference.
    *   **Filtering Data**.
    *   **Counting Unique Items**.

### Demonstrations of Python Data Structures and Operations

#### 1. String Operations (Length and Slicing)

```python
my_string = "Hello Python World!"
print(f"Original string: {my_string}")
print(f"Length of the string: {len(my_string)}")
print(f"First three characters: {my_string[:3]}")
```

#### 2. Sequence Data Types: Examples, Indexing, and Slicing

```python
# String
my_str = "Python"
print(f"String: {my_str}")
print(f"  Index 0: {my_str[0]}")
print(f"  Index -1: {my_str[-1]}")
print(f"  Slice [1:4]: {my_str[1:4]}\n")

# List
my_list = [10, 20, 30, 40, 50]
print(f"List: {my_list}")
print(f"  Index 0: {my_list[0]}")
print(f"  Index -2: {my_list[-2]}")
print(f"  Slice [2:]: {my_list[2:]}\n")

# Tuple
my_tuple = ('apple', 'banana', 'cherry', 'date')
print(f"Tuple: {my_tuple}")
print(f"  Index 1: {my_tuple[1]}")
print(f"  Index -1: {my_tuple[-1]}")
print(f"  Slice [:3]: {my_tuple[:3]}\n")

# Range
my_range = range(0, 10, 2) # 0, 2, 4, 6, 8
print(f"Range (as list): {list(my_range)}")
print(f"  Index 2: {my_range[2]}") # Accesses the 3rd element (4)
print(f"  Slice [1:4] (as list): {list(my_range[1:4])}") # (2, 4, 6)
```

#### 3. List vs. Tuple: Mutability Demonstration

```python
# List (Mutable)
mutable_list = [1, 2, 3]
print(f"Original list: {mutable_list}")
mutable_list.append(4)
print(f"List after append: {mutable_list}")
mutable_list[0] = 99
print(f"List after modification: {mutable_list}\n")

# Tuple (Immutable)
immutable_tuple = (1, 2, 3)
print(f"Original tuple: {immutable_tuple}")

# Attempting to modify a tuple will raise an error
try:
    immutable_tuple[0] = 99
except TypeError as e:
    print(f"Error attempting to modify tuple: {e}")

try:
    immutable_tuple.append(4)
except AttributeError as e:
    print(f"Error attempting to append to tuple: {e}")
```

#### 4. Python Arrays vs. Lists

```python
import array

# List (heterogeneous)
my_mixed_list = [1, "hello", 3.14]
print(f"List with mixed types: {my_mixed_list}\n")

# Array (homogeneous - type code 'i' for signed int)
# Refer to array module documentation for type codes (e.g., 'f' for float, 'd' for double)
my_int_array = array.array('i', [1, 2, 3, 4, 5])
print(f"Array of integers: {my_int_array}")

# Trying to add a different type to an array will raise an error
try:
    my_int_array.append(3.14)
except TypeError as e:
    print(f"Error attempting to append float to int array: {e}")

my_float_array = array.array('f', [1.1, 2.2, 3.3])
print(f"Array of floats: {my_float_array}")
```

#### 5. Dictionary Operations

```python
my_dict = {
    "name": "Alice",
    "age": 30,
    "city": "New York"
}
print(f"Original dictionary: {my_dict}")

# Accessing values
print(f"Name: {my_dict['name']}")
print(f"Age: {my_dict.get('age')}\n")

# Adding a new key-value pair
my_dict['occupation'] = 'Engineer'
print(f"Dictionary after adding 'occupation': {my_dict}\n")

# Modifying a value
my_dict['age'] = 31
print(f"Dictionary after modifying 'age': {my_dict}\n")

# Using a tuple as a key (immutable)
coordinates = (40.71, -74.00)
my_dict[coordinates] = "New York City Coordinates"
print(f"Dictionary with tuple key: {my_dict}")
print(f"Value for coordinates: {my_dict[(40.71, -74.00)]}")
```

#### 6. Set Operations

```python
# Creating a set
my_set = {1, 2, 3, 4, 5}
print(f"Initial set: {my_set}")

# Creating a set from a list with duplicates
list_with_duplicates = [1, 2, 2, 3, 4, 4, 5, 6]
unique_elements = set(list_with_duplicates)
print(f"Set from list with duplicates: {unique_elements}\n")

# Adding elements to a set
unique_elements.add(7)
print(f"Set after adding 7: {unique_elements}")
unique_elements.add(3) # No effect, 3 is already present
print(f"Set after adding 3 (no change): {unique_elements}\n")

# Removing elements
unique_elements.remove(1)
print(f"Set after removing 1: {unique_elements}\n")

# Membership testing (fast!)
print(f"Is 5 in the set? {5 in unique_elements}")
print(f"Is 10 in the set? {10 in unique_elements}\n")

# Set operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(f"Set A: {set_a}")
print(f"Set B: {set_b}\n")

print(f"Union (A | B): {set_a | set_b}")
print(f"Intersection (A & B): {set_a & set_b}")
print(f"Difference (A - B): {set_a - set_b}")
print(f"Symmetric Difference (A ^ B): {set_a ^ set_b}")
```
