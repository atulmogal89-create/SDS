### 1. What is Data Wrangling? Difference Between Data Wrangling and Data Preprocessing

*   **Data Wrangling** (also called Data Munging): The process of transforming and mapping raw data from one format into another to make it more appropriate and valuable for downstream analytics or machine learning tasks.
*   **Steps in Data Wrangling**:
    *   Discovering — understanding what data you have.
    *   Structuring — reshaping or reorganizing the data.
    *   Cleaning — fixing errors, handling missing values.
    *   Enriching — adding derived or external data.
    *   Validating — confirming data quality and consistency.
    *   Publishing — delivering clean data for analysis.

| Feature | Data Wrangling | Data Preprocessing |
|---|---|---|
| **Scope** | Broader — includes acquisition, transformation, enrichment | Narrower — focused on preparing data for ML models |
| **Goal** | Make data usable and analysis-ready | Make data suitable for model training |
| **Techniques** | Merging, reshaping, renaming, filtering | Normalization, encoding, scaling, splitting |
| **Stage** | Early in the data pipeline | Just before model building |
| **Tools** | pandas, OpenRefine, Trifacta | scikit-learn, pandas, NumPy |

### 2. Importance of Exploratory Data Analysis (EDA) in Pandas

*   **Definition**: EDA is an approach to analyzing datasets to summarize their main characteristics, often using visual methods, before applying formal modeling.
*   **Why it matters**:
    *   **Understand the data**: Reveals structure, distributions, and relationships between variables.
    *   **Detect anomalies**: Identifies outliers, missing values, and data entry errors early.
    *   **Test assumptions**: Validates assumptions required by statistical models.
    *   **Feature selection**: Highlights which columns are most relevant to the analysis.
    *   **Guides cleaning**: Helps decide which cleaning steps are necessary.
*   **Key pandas tools for EDA**:
    *   `df.head()`, `df.tail()` — view first/last rows.
    *   `df.info()` — column types and non-null counts.
    *   `df.describe()` — statistical summary (mean, std, min, max, quartiles).
    *   `df.value_counts()` — frequency of unique values.
    *   `df.corr()` — correlation matrix between numerical columns.

### 3. Main Differences Between Series, DataFrame, and Panel

| Feature | Series | DataFrame | Panel |
|---|---|---|---|
| **Dimensions** | 1D (one axis) | 2D (rows × columns) | 3D (deprecated in pandas 0.25+) |
| **Structure** | Single column of data with an index | Table with labeled rows and columns | Collection of DataFrames |
| **Creation** | `pd.Series([1, 2, 3])` | `pd.DataFrame({...})` | `pd.Panel(...)` (deprecated) |
| **Use case** | Single variable / column | Tabular / structured data | Multi-dimensional data (now use MultiIndex or xarray) |
| **Index** | One index | Row index + column labels | items, major_axis, minor_axis |

*   **Series**: A one-dimensional labeled array capable of holding any data type.
*   **DataFrame**: A two-dimensional, size-mutable, tabular data structure with labeled axes. Most commonly used pandas object.
*   **Panel**: A three-dimensional data structure (deprecated). Replaced by `pd.MultiIndex` DataFrames or the `xarray` library.

### 4. Explain the Use of isnull(), notnull(), dropna(), fillna()

*   **`isnull()`**: Returns a DataFrame/Series of the same shape with `True` where values are `NaN` (missing) and `False` elsewhere. Used to detect missing values.
    *   Example: `df.isnull().sum()` gives count of missing values per column.

*   **`notnull()`**: The inverse of `isnull()`. Returns `True` where values are NOT missing. Used to filter or count present values.
    *   Example: `df[df['Age'].notnull()]` returns rows where Age is not missing.

*   **`dropna()`**: Removes rows or columns that contain missing values.
    *   `df.dropna()` — drops any row with at least one NaN.
    *   `df.dropna(axis=1)` — drops columns with NaN.
    *   `df.dropna(thresh=2)` — keeps rows with at least 2 non-NaN values.

*   **`fillna()`**: Fills missing values with a specified value or method.
    *   `df.fillna(0)` — fills all NaN with 0.
    *   `df['col'].fillna(df['col'].mean())` — fills with column mean.
    *   `df.fillna(method='ffill')` — forward fill (propagate last valid value forward).
    *   `df.fillna(method='bfill')` — backward fill.

### 5. What is Column Transformation and Why is it Important?

*   **Definition**: Column transformation refers to applying operations on one or more columns to create new columns, modify existing ones, or change their representation to better serve the analysis.
*   **Types of Column Transformation**:
    *   **Mathematical**: `df['Salary_k'] = df['Salary'] / 1000` — scale values.
    *   **String operations**: `.str.upper()`, `.str.replace()` — clean text data.
    *   **Type casting**: `.astype()` — convert column types.
    *   **Binning**: `pd.cut()`, `pd.qcut()` — group continuous values into categories.
    *   **apply() / map() / lambda**: Apply custom functions element-wise.
    *   **Encoding**: Convert categorical to numerical (Label Encoding, One-Hot Encoding).
*   **Why it is important**:
    *   Makes data compatible with algorithms that require specific formats.
    *   Creates meaningful features that improve analysis and model accuracy.
    *   Normalizes skewed distributions for better statistical validity.
    *   Simplifies complex columns into more interpretable forms.

### Demonstrations of Data Wrangling with pandas

#### 1. Creating a Series and a DataFrame

```python
import pandas as pd
import numpy as np

# --- Series ---
salary_series = pd.Series([55000, 72000, 48000, 61000, 83000],
                           index=['Alice', 'Bob', 'Charlie', 'Eva', 'Frank'],
                           name='Salary')
print("--- pandas Series ---")
print(salary_series)
print(f"\nType: {type(salary_series)}")
print(f"dtype: {salary_series.dtype}")

# --- DataFrame ---
data = {
    'Name':       ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Heidi', 'Ivan', 'Judy'],
    'Age':        [28, 35, np.nan, 45, 29, 38, np.nan, 31, 27, 33],
    'Gender':     ['Female', 'Male', 'Male', 'Male', 'female', 'Male', 'Female', ' FEMALE', 'Male', 'Female'],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'Marketing', 'IT', 'Finance', 'HR', None, 'Marketing'],
    'Salary':     [55000, 72000, 48000, np.nan, 61000, 83000, 47000, 52000, 44000, np.nan],
    'Experience': [3, 10, 5, 20, 4, 15, 6, 8, 2, 9]
}

df = pd.DataFrame(data)
print("\n--- pandas DataFrame ---")
print(df)
print(f"\nShape: {df.shape}")
```

#### 2. EDA — Exploratory Data Analysis

```python
print("--- df.info() ---")
df.info()

print("\n--- df.describe() ---")
print(df.describe())

print("\n--- df.head(5) ---")
print(df.head(5))

print("\n--- Value Counts: Department ---")
print(df['Department'].value_counts())

print("\n--- Correlation Matrix ---")
print(df[['Age', 'Salary', 'Experience']].corr())
```

#### 3. isnull() and notnull()

```python
print("--- isnull() : Missing value map ---")
print(df.isnull())

print("\n--- Missing value count per column ---")
print(df.isnull().sum())

print("\n--- notnull() : Rows where Salary is present ---")
print(df[df['Salary'].notnull()])

print("\n--- notnull() : Non-null count per column ---")
print(df.notnull().sum())
```

#### 4. dropna() — Removing Missing Values

```python
# Drop rows with ANY missing value
df_dropped_any = df.dropna()
print("--- dropna() : Drop rows with any NaN ---")
print(df_dropped_any)
print(f"Rows remaining: {len(df_dropped_any)}")

# Drop rows where ALL values are missing
df_dropped_all = df.dropna(how='all')
print(f"\n--- dropna(how='all') : Rows remaining: {len(df_dropped_all)} ---")

# Drop rows with NaN only in specific columns
df_dropped_salary = df.dropna(subset=['Salary'])
print("\n--- dropna(subset=['Salary']) : Drop rows missing Salary ---")
print(df_dropped_salary)
```

#### 5. fillna() — Filling Missing Values

```python
df_filled = df.copy()

# Fill Age with median
df_filled['Age'] = df_filled['Age'].fillna(df_filled['Age'].median())
print(f"Filled Age with median: {df['Age'].median()}")

# Fill Salary with mean
df_filled['Salary'] = df_filled['Salary'].fillna(round(df_filled['Salary'].mean(), 2))
print(f"Filled Salary with mean: {round(df['Salary'].mean(), 2)}")

# Fill Department with mode (most frequent)
df_filled['Department'] = df_filled['Department'].fillna(df_filled['Department'].mode()[0])
print(f"Filled Department with mode: {df['Department'].mode()[0]}")

print("\n--- Missing values after fillna() ---")
print(df_filled.isnull().sum())

print("\n--- DataFrame after fillna() ---")
print(df_filled)
```

#### 6. Column Transformation — apply(), map(), lambda

```python
df_wrangled = df_filled.copy()

# Standardize Gender column
df_wrangled['Gender'] = df_wrangled['Gender'].str.strip().str.title()
print("--- Gender after standardization ---")
print(df_wrangled['Gender'].unique())

# Create a new column: Salary in thousands
df_wrangled['Salary_K'] = df_wrangled['Salary'].apply(lambda x: round(x / 1000, 1))
print("\n--- New column: Salary_K (in thousands) ---")
print(df_wrangled[['Name', 'Salary', 'Salary_K']])

# Classify employees as 'Junior' or 'Senior' based on Experience
df_wrangled['Level'] = df_wrangled['Experience'].map(lambda x: 'Senior' if x >= 10 else 'Junior')
print("\n--- New column: Level (based on Experience) ---")
print(df_wrangled[['Name', 'Experience', 'Level']])
```

#### 7. Binning with pd.cut() and Renaming Columns

```python
# Bin Age into categories
df_wrangled['Age_Group'] = pd.cut(
    df_wrangled['Age'],
    bins=[20, 30, 40, 50],
    labels=['20-30', '30-40', '40-50']
)
print("--- Age Binning with pd.cut() ---")
print(df_wrangled[['Name', 'Age', 'Age_Group']])

# Rename columns for clarity
df_wrangled = df_wrangled.rename(columns={
    'Salary_K': 'Salary (K)',
    'Age_Group': 'Age Group',
    'Level': 'Seniority'
})
print("\n--- Columns after renaming ---")
print(df_wrangled.columns.tolist())
```

#### 8. Filtering, Sorting, and Grouping

```python
# Filtering: Senior employees with Salary > 60000
senior_high = df_wrangled[(df_wrangled['Seniority'] == 'Senior') & (df_wrangled['Salary'] > 60000)]
print("--- Senior employees earning > 60,000 ---")
print(senior_high[['Name', 'Department', 'Salary', 'Experience', 'Seniority']])

# Sorting: by Salary descending
df_sorted = df_wrangled.sort_values(by='Salary', ascending=False)
print("\n--- Data sorted by Salary (descending) ---")
print(df_sorted[['Name', 'Department', 'Salary']].reset_index(drop=True))

# Grouping: Department-wise average Salary and Experience
print("\n--- Department-wise Average Salary and Experience ---")
print(df_wrangled.groupby('Department')[['Salary', 'Experience']].mean().round(2))

print("\n--- Final Wrangled Dataset ---")
print(df_wrangled)
```
