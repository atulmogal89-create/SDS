### 1. Importance of Data Visualization in Data Analysis and EDA

*   **Definition**: Data visualization is the graphical representation of information and data using visual elements like charts, graphs, and plots to make patterns, trends, and outliers easy to see.
*   **Importance in Data Analysis**:
    *   **Pattern Recognition**: Visuals reveal trends, cycles, and correlations that are invisible in raw tables.
    *   **Outlier Detection**: Box plots and scatter plots immediately expose anomalous data points.
    *   **Distribution Understanding**: Histograms show how data is spread and whether it is skewed or normal.
    *   **Communication**: Complex findings are conveyed quickly and clearly to non-technical audiences.
    *   **Decision Support**: Dashboards and charts enable faster, data-driven decisions.
*   **Role in EDA**:
    *   **Univariate Analysis**: Histograms and box plots reveal the distribution and spread of a single variable.
    *   **Bivariate Analysis**: Scatter plots and line plots show relationships between two variables.
    *   **Multivariate Analysis**: Pair plots and heatmaps reveal interactions among multiple variables simultaneously.
    *   **Missing Value Visualization**: Heatmaps of `isnull()` help identify missing data patterns.
    *   **Feature Relationships**: Correlation heatmaps guide feature selection for machine learning.

### 2. Difference Between Matplotlib and Other Visualization Libraries

| Feature | Matplotlib | Seaborn | Plotly | Bokeh |
|---|---|---|---|---|
| **Type** | Static plots | Statistical plots (built on Matplotlib) | Interactive plots | Interactive web plots |
| **Ease of use** | Verbose, fine-grained control | High-level, concise syntax | Moderate | Moderate |
| **Interactivity** | None (static) | None (static) | High (hover, zoom, click) | High |
| **Statistical plots** | Manual | Built-in (violin, pair, heatmap) | Some | Limited |
| **Customization** | Very high | Moderate | High | High |
| **Best for** | Publication-quality static charts | Statistical EDA | Web dashboards | Web applications |

*   **Matplotlib** is the most foundational and gives pixel-level control over every element of a figure.
*   **Seaborn** wraps Matplotlib with a high-level API and beautiful default themes, ideal for statistical visualization.
*   **Plotly** and **Bokeh** produce interactive, web-ready charts but require more setup for publication output.

### 3. Purpose of the pyplot Module in Matplotlib

*   `matplotlib.pyplot` is a collection of functions that make Matplotlib work like MATLAB's plotting environment.
*   It provides a **state-machine interface** — each function call changes some part of the current figure or axes.
*   **Key roles of pyplot**:
    *   `plt.figure()` — creates a new figure canvas.
    *   `plt.subplot()` / `plt.subplots()` — creates multiple axes in one figure.
    *   `plt.plot()`, `plt.bar()`, `plt.scatter()` etc. — draws plots on the current axes.
    *   `plt.title()`, `plt.xlabel()`, `plt.ylabel()` — adds labels and titles.
    *   `plt.legend()` — adds a legend.
    *   `plt.grid()` — toggles a background grid.
    *   `plt.show()` — renders and displays the figure.
    *   `plt.savefig()` — saves the figure to a file (PNG, PDF, SVG, etc.).
*   Without `pyplot`, creating plots requires manually managing `Figure` and `Axes` objects in an object-oriented style.

### 4. Process of Creating a Basic Line Plot

*   A line plot connects data points in the order they appear, making it ideal for showing trends over time or continuous variables.
*   **Steps**:
    1.  **Import pyplot**: `import matplotlib.pyplot as plt`
    2.  **Prepare data**: Define x and y values (lists or NumPy arrays).
    3.  **Create the plot**: `plt.plot(x, y)` draws the line.
    4.  **Customize**: Add `color`, `linestyle`, `linewidth`, `marker` parameters.
    5.  **Add labels**: `plt.title()`, `plt.xlabel()`, `plt.ylabel()`.
    6.  **Add legend**: `plt.legend()` if multiple lines are plotted.
    7.  **Display**: `plt.show()` renders the figure.
*   **Example**:
    ```python
    import matplotlib.pyplot as plt
    x = [1, 2, 3, 4, 5]
    y = [10, 25, 18, 40, 33]
    plt.plot(x, y, color='blue', marker='o', linestyle='--', linewidth=2)
    plt.title('Basic Line Plot')
    plt.xlabel('X-axis')
    plt.ylabel('Y-axis')
    plt.grid(True)
    plt.show()
    ```

### 5. Importance of Customization and Available Options in Matplotlib

*   **Why Customization Matters**:

    *   Highlights **specific data points** or regions of interest using annotations.

### Demonstrations of Data Visualization with Matplotlib and Seaborn

#### 0. Dataset Preparation

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set global style
sns.set_style('whitegrid')
sns.set_palette('deep')
np.random.seed(42)

# Create a sample Employee dataset
n = 80
departments = np.random.choice(['HR', 'IT', 'Finance', 'Marketing'], size=n)
experience  = np.random.randint(1, 20, size=n)
salary      = 30000 + experience * 2500 + np.random.randint(-5000, 5000, size=n)
age         = 22 + experience + np.random.randint(0, 5, size=n)
performance = np.round(np.clip(experience * 0.3 + np.random.randn(n) * 1.5 + 4, 1, 10), 1)

df = pd.DataFrame({
    'Department' : departments,
    'Experience' : experience,
    'Salary'     : salary,
    'Age'        : age,
    'Performance': performance
})

print("--- Dataset Preview ---")
print(df.head(10))
print(f"\nShape: {df.shape}")
print("\n--- Statistical Summary ---")
print(df.describe())
````

#### 1. Scatter Plot — Experience vs Salary

````python
# Color-code by Department
dept_colors = {'HR': 'steelblue', 'IT': 'tomato', 'Finance': 'seagreen', 'Marketing': 'darkorange'}
colors = df['Department'].map(dept_colors)

plt.figure(figsize=(9, 5))
scatter = plt.scatter(
    df['Experience'], df['Salary'],
    c=colors, alpha=0.75, edgecolors='white', linewidths=0.5, s=80
)

# Manual legend for departments
for dept, color in dept_colors.items():
    plt.scatter([], [], color=color, label=dept, s=60)

plt.title('Scatter Plot: Experience vs Salary (by Department)', fontsize=14, fontweight='bold')
plt.xlabel('Years of Experience', fontsize=12)
plt.ylabel('Salary (INR)', fontsize=12)
plt.legend(title='Department', loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
print("Observation: Salary shows a positive correlation with Experience across all departments.")
````

#### 2. Line Plot — Average Salary Trend by Experience

````python
# Group by Experience and compute mean salary
avg_salary = df.groupby('Experience')['Salary'].mean().reset_index()

plt.figure(figsize=(10, 5))
plt.plot(
    avg_salary['Experience'], avg_salary['Salary'],
    color='royalblue', linewidth=2.5, marker='o',
    markersize=7, markerfacecolor='white', markeredgewidth=2,
    label='Avg Salary'
)

# Add a trend reference line
z = np.polyfit(avg_salary['Experience'], avg_salary['Salary'], 1)
p = np.poly1d(z)
plt.plot(avg_salary['Experience'], p(avg_salary['Experience']),
         color='tomato', linewidth=1.8, linestyle='--', label='Trend Line')

plt.title('Line Plot: Average Salary Trend by Years of Experience', fontsize=14, fontweight='bold')
plt.xlabel('Years of Experience', fontsize=12)
plt.ylabel('Average Salary (INR)', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
print("Observation: Average salary increases steadily with experience, with some variation at higher levels.")
````

#### 3. Bar Plot — Average Salary by Department

````python
dept_avg = df.groupby('Department')['Salary'].mean().sort_values(ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# --- Matplotlib Bar Plot ---
bar_colors = ['steelblue', 'seagreen', 'darkorange', 'tomato']
bars = axes[0].bar(dept_avg.index, dept_avg.values, color=bar_colors, edgecolor='white', width=0.55)
for bar in bars:
    axes[0].text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 400,
        f"{bar.get_height():,.0f}",
        ha='center', va='bottom', fontsize=10, fontweight='bold'
    )
axes[0].set_title('Matplotlib: Avg Salary by Department', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Department', fontsize=11)
axes[0].set_ylabel('Average Salary (INR)', fontsize=11)
axes[0].grid(axis='y', linestyle='--', alpha=0.5)

# --- Seaborn Bar Plot ---
sns.barplot(data=df, x='Department', y='Salary', ax=axes[1],
            palette='deep', capsize=0.1, errwidth=1.5)
axes[1].set_title('Seaborn: Avg Salary by Department (with CI)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Department', fontsize=11)
axes[1].set_ylabel('Average Salary (INR)', fontsize=11)

plt.tight_layout()
plt.show()
print("Observation: IT department shows the highest average salary. Seaborn adds confidence intervals automatically.")
````

#### 4. Histogram — Salary Distribution

````python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# --- Matplotlib Histogram ---
axes[0].hist(df['Salary'], bins=15, color='steelblue', edgecolor='white', alpha=0.85)
axes[0].axvline(df['Salary'].mean(), color='tomato', linewidth=2,
                linestyle='--', label=f"Mean: {df['Salary'].mean():,.0f}")
axes[0].axvline(df['Salary'].median(), color='seagreen', linewidth=2,
                linestyle='-.',  label=f"Median: {df['Salary'].median():,.0f}")
axes[0].set_title('Matplotlib: Salary Distribution', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Salary (INR)', fontsize=11)
axes[0].set_ylabel('Frequency', fontsize=11)
axes[0].legend(fontsize=10)
axes[0].grid(axis='y', linestyle='--', alpha=0.5)

# --- Seaborn Histogram with KDE ---
sns.histplot(df['Salary'], bins=15, kde=True, color='darkorange',
             edgecolor='white', alpha=0.8, ax=axes[1])
axes[1].set_title('Seaborn: Salary Distribution with KDE', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Salary (INR)', fontsize=11)
axes[1].set_ylabel('Count', fontsize=11)

plt.tight_layout()
plt.show()
print("Observation: Salary distribution is roughly bell-shaped with a slight right skew. KDE curve shows the smooth density estimate.")
````

#### 5. Box Plot — Salary Distribution by Department

````python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# --- Matplotlib Box Plot ---
dept_list = ['HR', 'IT', 'Finance', 'Marketing']
salary_by_dept = [df[df['Department'] == d]['Salary'].values for d in dept_list]

bp = axes[0].boxplot(salary_by_dept, labels=dept_list, patch_artist=True, notch=False)
box_colors = ['steelblue', 'tomato', 'seagreen', 'darkorange']
for patch, color in zip(bp['boxes'], box_colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.75)
for median in bp['medians']:
    median.set_color('black')
    median.set_linewidth(2)
axes[0].set_title('Matplotlib: Salary Box Plot by Department', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Department', fontsize=11)
axes[0].set_ylabel('Salary (INR)', fontsize=11)
axes[0].grid(axis='y', linestyle='--', alpha=0.5)

# --- Seaborn Box Plot ---
sns.boxplot(data=df, x='Department', y='Salary', palette='deep',
            ax=axes[1], linewidth=1.5)
sns.stripplot(data=df, x='Department', y='Salary', color='black',
              alpha=0.3, size=3.5, ax=axes[1], jitter=True)
axes[1].set_title('Seaborn: Box + Strip Plot by Department', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Department', fontsize=11)
axes[1].set_ylabel('Salary (INR)', fontsize=11)

plt.tight_layout()
plt.show()
print("Observation: IT shows the widest salary spread. Outliers are visible in Finance and Marketing.")
````

#### 6. Pair Plot — Multivariate Relationships (Seaborn)

````python
# Select numerical columns for pair plot
pair_df = df[['Experience', 'Salary', 'Age', 'Performance', 'Department']]

pair_grid = sns.pairplot(
    pair_df,
    hue='Department',
    palette='deep',
    diag_kind='kde',
    plot_kws={'alpha': 0.6, 'edgecolor': 'none', 's': 40},
    diag_kws={'linewidth': 2}
)
pair_grid.fig.suptitle('Pair Plot: Multivariate Relationships by Department',
                        y=1.02, fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
print("Observation: Strong positive correlation between Experience and Salary. Age and Experience move together as expected. KDE diagonals show department-wise distributions.")
````

#### 7. Bonus — Correlation Heatmap (Seaborn)

````python
plt.figure(figsize=(7, 5))
corr_matrix = df[['Experience', 'Salary', 'Age', 'Performance']].corr()

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.8,
    linecolor='white',
    vmin=-1, vmax=1,
    square=True,
    annot_kws={'size': 12}
)
plt.title('Correlation Heatmap: Numerical Features', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
print("Observation: Experience and Salary have the highest correlation (close to 1.0). Performance has a moderate positive correlation with Experience.")
````
