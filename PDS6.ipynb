### 1. What is Web Scraping? Role of the Requests Library

*   **Web Scraping**: The automated process of extracting data from websites by sending HTTP requests, receiving HTML responses, and parsing the content to retrieve useful information.
*   **Use Cases**:
    *   Price monitoring and comparison across e-commerce sites.
    *   Collecting news headlines or articles for NLP research.
    *   Aggregating job listings, real estate data, or sports statistics.
    *   Academic research requiring large-scale data collection.
    *   Building datasets when no official API is available.

*   **Role of the `requests` Library**:
    *   `requests` is a Python HTTP library that allows Python programs to send HTTP/HTTPS requests to web servers and receive responses — just like a web browser does.
    *   **Key functions**:
        *   `requests.get(url)` — sends a GET request and returns a `Response` object.
        *   `response.status_code` — HTTP status (200 = OK, 404 = Not Found, 403 = Forbidden).
        *   `response.text` — raw HTML content of the page as a string.
        *   `response.content` — raw bytes (for images, PDFs).
        *   `response.headers` — response headers from the server.
        *   `requests.get(url, headers={...})` — pass custom headers (e.g., User-Agent) to mimic a real browser.
    *   Without `requests`, we cannot download the raw HTML; BeautifulSoup alone cannot fetch web pages.

### 2. What is BeautifulSoup? How Does it Help in Parsing HTML?

*   **BeautifulSoup**: A Python library (`bs4`) for parsing HTML and XML documents. It creates a parse tree from the raw HTML, which can be navigated, searched, and modified easily.
*   **How it helps**:
    *   Converts the raw HTML string (from `requests`) into a structured, navigable Python object.
    *   Provides intuitive methods to locate elements by tag name, class, id, or attributes.
    *   Handles badly formed or malformed HTML gracefully.
    *   Extracts text, links, table data, images, and any HTML element cleanly.
*   **Parsers supported by BeautifulSoup**:

| Parser | Usage | Speed | Notes |
|---|---|---|---|
| `html.parser` | `BeautifulSoup(html, 'html.parser')` | Moderate | Built into Python, no extra install |
| `lxml` | `BeautifulSoup(html, 'lxml')` | Very Fast | Requires `pip install lxml` |
| `html5lib` | `BeautifulSoup(html, 'html5lib')` | Slow | Most lenient, handles broken HTML |

*   **Workflow**:
    ```
    URL → requests.get() → HTML text → BeautifulSoup() → Parse tree → find/find_all → Data
    ```

### 3. Differentiate Between Web Scraping and Web Crawling

| Feature | Web Scraping | Web Crawling |
|---|---|---|
| **Definition** | Extracting specific data from a web page | Systematically browsing the web to discover and index pages |
| **Scope** | Targeted — specific pages or elements | Broad — follows links across many pages/sites |
| **Output** | Structured data (CSV, JSON, DB) | URLs, page index, or map of the web |
| **Purpose** | Data collection for analysis | Building a search index or site map |
| **Tools** | BeautifulSoup, Scrapy, Selenium | Scrapy, Googlebot, custom spiders |
| **Example** | Extracting product prices from Amazon | Google indexing every page on the internet |
| **Volume** | Small to medium (targeted pages) | Large scale (millions of pages) |

*   **Analogy**: Web crawling is like a librarian cataloguing every book in a library (finding what exists). Web scraping is like reading and extracting specific chapters from selected books (extracting content).

### 4. Commonly Used Methods in BeautifulSoup

*   **`find(tag, attrs)`**: Returns the **first** matching tag.
    *   `soup.find('h1')` — first `<h1>` tag.
    *   `soup.find('div', class_='price')` — first `<div>` with class `price`.

*   **`find_all(tag, attrs)`**: Returns a **list of all** matching tags.
    *   `soup.find_all('a')` — all anchor/link tags.
    *   `soup.find_all('p', limit=5)` — first 5 `<p>` tags.

*   **`.text` / `.get_text()`**: Extracts visible text content from a tag, stripping all HTML.
    *   `tag.get_text(strip=True)` — text with surrounding whitespace removed.

*   **`.get(attribute)`**: Retrieves the value of a specific HTML attribute.
    *   `tag.get('href')` — extracts the URL from an `<a>` tag.
    *   `tag.get('src')` — extracts the image source from an `<img>` tag.

*   **`select(css_selector)`**: Uses CSS selectors to find elements (similar to jQuery).
    *   `soup.select('table.data > tr')` — all `<tr>` rows inside a table with class `data`.
    *   `soup.select('#main-content h2')` — all `<h2>` inside element with id `main-content`.

*   **`select_one(css_selector)`**: Returns the first match for a CSS selector.

*   **`.parent`, `.children`, `.siblings`**: Navigate the parse tree relationally.

*   **`.attrs`**: Returns a dictionary of all attributes of a tag.
    *   `tag.attrs` → `{'class': ['product'], 'id': 'item-1'}`

### 5. Ethical and Legal Considerations in Web Scraping

*   **Respect `robots.txt`**: Every website has a `/robots.txt` file specifying which pages may or may not be scraped. Ethical scrapers always check and comply with this file.
*   **Terms of Service (ToS)**: Many websites explicitly prohibit scraping in their Terms of Service. Violating ToS can result in account bans, legal action, or IP blocking.
*   **Rate Limiting**: Sending too many requests in a short time can overload a server (equivalent to a Denial-of-Service attack). Always add delays between requests (`time.sleep()`).
*   **Data Privacy**: Scraping personal data (names, emails, phone numbers) without consent may violate privacy laws such as GDPR (Europe) and IT Act (India).
*   **Copyright**: Web content (articles, images, code) is protected by copyright. Reproducing or republishing scraped content without permission is illegal.
*   **Authentication Bypass**: Scraping pages behind login walls without authorization is illegal under the Computer Fraud and Abuse Act (CFAA, USA) and similar laws.
*   **Best Practices**:
    *   Always check `robots.txt` and ToS before scraping.
    *   Use official APIs when available — they are faster, legal, and structured.
    *   Identify your scraper honestly via a User-Agent header.
    *   Never scrape and commercially distribute copyrighted data.
    *   Add polite delays (`time.sleep(1–3 seconds)`) between requests.

#### 1. Parsing a Local HTML String with BeautifulSoup

````python
# Simulated HTML document (as if fetched from a web page)
sample_html = """
<html>
  <head><title>Student Results Portal</title></head>
  <body>
    <h1 id='main-heading'>Semester Results</h1>
    <p class='info'>Department: Computer Science</p>
    <p class='info'>Academic Year: 2024-25</p>

    <table id='results-table'>
      <tr><th>Name</th><th>Roll No</th><th>Marks</th><th>Grade</th></tr>
      <tr><td>Alice</td><td>01</td><td>88</td><td>A</td></tr>
      <tr><td>Bob</td><td>02</td><td>74</td><td>B</td></tr>
      <tr><td>Charlie</td><td>03</td><td>91</td><td>A+</td></tr>
      <tr><td>David</td><td>04</td><td>65</td><td>C</td></tr>
      <tr><td>Eva</td><td>05</td><td>79</td><td>B+</td></tr>
    </table>

    <h2>Useful Links</h2>
    <ul>
      <li><a href='https://www.python.org'>Python Official Site</a></li>
      <li><a href='https://www.numpy.org'>NumPy Documentation</a></li>
      <li><a href='https://pandas.pydata.org'>Pandas Documentation</a></li>
    </ul>
  </body>
</html>
"""

# Parse the HTML
soup = BeautifulSoup(sample_html, 'html.parser')

print("--- Parsed HTML (prettified preview) ---")
print(soup.prettify()[:600], "...\n")

print(f"Page Title  : {soup.title.text}")
print(f"H1 Heading  : {soup.find('h1').text}")
print(f"H2 Heading  : {soup.find('h2').text}")
````

#### 2. find() and find_all() Methods

````python
# --- find(): first matching tag ---
first_p = soup.find('p')
print(f"find('p') — First <p> tag text : '{first_p.text}'")

heading = soup.find('h1', id='main-heading')
print(f"find('h1', id='main-heading')  : '{heading.text}'")

# --- find_all(): list of all matching tags ---
all_p = soup.find_all('p', class_='info')
print(f"\nfind_all('p', class_='info') — {len(all_p)} tags found:")
for p in all_p:
    print(f"  {p.text}")

# --- All <td> tags ---
all_td = soup.find_all('td')
print(f"\nfind_all('td') — {len(all_td)} <td> tags found")

# --- All <a> tags (links) ---
all_links = soup.find_all('a')
print(f"\nfind_all('a') — {len(all_links)} links found:")
for link in all_links:
    print(f"  Text: {link.text:30s}  |  href: {link.get('href')}")
````

#### 3. Scraping Table Data into a pandas DataFrame

````python
table = soup.find('table', id='results-table')

# Extract headers
headers = [th.get_text(strip=True) for th in table.find_all('th')]
print(f"Headers: {headers}")

# Extract rows (skip header row)
rows = []
for tr in table.find_all('tr')[1:]:       # skip the <th> row
    cells = [td.get_text(strip=True) for td in tr.find_all('td')]
    if cells:
        rows.append(cells)

# Build DataFrame
df_results = pd.DataFrame(rows, columns=headers)
df_results['Marks'] = df_results['Marks'].astype(int)

print("\n--- Scraped Table as DataFrame ---")
print(df_results)
print(f"\nAverage Marks : {df_results['Marks'].mean():.1f}")
print(f"Top Scorer    : {df_results.loc[df_results['Marks'].idxmax(), 'Name']} ({df_results['Marks'].max()})")
````

#### 4. CSS Selectors with select() and select_one()

````python
# select_one() — CSS selector for single element
title = soup.select_one('head title')
print(f"select_one('head title')   : '{title.text}'")

main_h1 = soup.select_one('h1#main-heading')
print(f"select_one('h1#main-heading'): '{main_h1.text}'")

# select() — CSS selector for multiple elements
info_paras = soup.select('p.info')
print(f"\nselect('p.info') — {len(info_paras)} elements:")
for p in info_paras:
    print(f"  {p.text}")

# Select all table rows from the results table
table_rows = soup.select('table#results-table tr')
print(f"\nselect('table#results-table tr') — {len(table_rows)} rows (including header)")

# Select all links inside <li> tags
li_links = soup.select('ul li a')
print(f"\nselect('ul li a') — {len(li_links)} links:")
for a in li_links:
    print(f"  {a.text}  →  {a.get('href')}")
````

#### 5. Navigating the Parse Tree — .parent, .children, .attrs

````python
# .attrs — all attributes of a tag
table_tag = soup.find('table')
print(f".attrs of <table>: {table_tag.attrs}")

h1_tag = soup.find('h1')
print(f".attrs of <h1>  : {h1_tag.attrs}")

# .parent — navigate up the tree
first_td = soup.find('td')
print(f"\nFirst <td> text            : '{first_td.text}'")
print(f"Its parent tag             : <{first_td.parent.name}>")
print(f"Grandparent tag            : <{first_td.parent.parent.name}>")

# .children — direct children of a tag
ul_tag = soup.find('ul')
print("\nChildren of <ul>:")
for child in ul_tag.children:
    if child.name:   # skip NavigableString whitespace nodes
        print(f"  <{child.name}>: {child.get_text(strip=True)}")

# get_text() with separator
body_text = soup.body.get_text(separator=' | ', strip=True)
print(f"\nAll body text (first 300 chars):\n{body_text[:300]}...")
````

#### 6. Live Web Scraping — Quotes from books.toscrape.com

````python
# books.toscrape.com is a legal sandbox site made for scraping practice
URL = 'http://books.toscrape.com/catalogue/page-1.html'

headers = {
    'User-Agent': 'Mozilla/5.0 (educational scraper; contact: student@example.com)'
}

try:
    response = requests.get(URL, headers=headers, timeout=10)
    print(f"Status Code : {response.status_code}")
    print(f"Content-Type: {response.headers.get('Content-Type', 'N/A')}")

    if response.status_code == 200:
        soup_live = BeautifulSoup(response.text, 'html.parser')

        # Scrape book titles, prices, and ratings
        books = soup_live.find_all('article', class_='product_pod')
        print(f"\nBooks found on page 1: {len(books)}")

        rating_map = {'One': 1, 'Two': 2, 'Three': 3, 'Four': 4, 'Five': 5}
        book_data = []
        for book in books:
            title  = book.h3.a.get('title')
            price  = book.find('p', class_='price_color').get_text(strip=True)
            rating_word = book.find('p', class_='star-rating')['class'][1]
            rating = rating_map.get(rating_word, 0)
            avail  = book.find('p', class_='instock').get_text(strip=True)
            book_data.append({'Title': title, 'Price': price,
                              'Rating': rating, 'Availability': avail})

        df_books = pd.DataFrame(book_data)
        print("\n--- Scraped Books (Page 1) ---")
        print(df_books.to_string(index=False))

        print(f"\nAverage Rating : {df_books['Rating'].mean():.2f} / 5")
        print(f"Highest Rated  : {df_books.loc[df_books['Rating'].idxmax(), 'Title']}")

        time.sleep(1)   # polite delay
    else:
        print("Failed to retrieve the page.")

except requests.exceptions.ConnectionError:
    print("[Offline] Could not connect. Run this cell when connected to the internet.")
except Exception as e:
    print(f"Error: {e}")
````

#### 7. Checking robots.txt and Extracting All Links from a Page

````python
BASE_URL = 'http://books.toscrape.com'

try:
    # --- Check robots.txt (ethical practice) ---
    robots_resp = requests.get(f"{BASE_URL}/robots.txt", timeout=10)
    print("--- robots.txt ---")
    print(robots_resp.text if robots_resp.status_code == 200 else "Not found")

    # --- Extract all internal hyperlinks from the page ---
    page_resp = requests.get(BASE_URL, headers=headers, timeout=10)
    if page_resp.status_code == 200:
        soup_links = BeautifulSoup(page_resp.text, 'html.parser')
        all_a_tags = soup_links.find_all('a', href=True)

        # Build full URLs and deduplicate
        links = []
        seen  = set()
        for a in all_a_tags:
            href = a['href']
            if href.startswith('http'):
                full_url = href
            else:
                full_url = BASE_URL + '/' + href.lstrip('./')
            if full_url not in seen:
                seen.add(full_url)
                links.append({'Text': a.get_text(strip=True), 'URL': full_url})

        df_links = pd.DataFrame(links)
        print(f"\n--- Unique Links Found: {len(df_links)} ---")
        print(df_links.head(15).to_string(index=False))

        time.sleep(1)

except requests.exceptions.ConnectionError:
    print("[Offline] Connect to the internet to run this cell.")
except Exception as e:
    print(f"Error: {e}")
````
