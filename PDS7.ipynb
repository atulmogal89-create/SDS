### 1. Difference Between Binomial, Poisson, and Normal Distributions

* **Binomial Distribution**:
    * Models the number of **successes in a fixed number of independent trials**.
    * Each trial has exactly two outcomes: success or failure.
    * Parameters: `n` (number of trials), `p` (probability of success).
    * Discrete distribution. Example: Number of heads in 10 coin flips.

* **Poisson Distribution**:
    * Models the number of **events occurring in a fixed interval** of time or space.
    * Events occur independently and at a constant average rate.
    * Parameter: `lambda` (average number of events per interval).
    * Discrete distribution. Example: Number of customer arrivals per hour.

* **Normal Distribution**:
    * Models **continuous data** that clusters symmetrically around a mean.
    * Bell-shaped, symmetric curve defined by mean and standard deviation.
    * Continuous distribution. Example: Heights, weights, exam scores.

| Feature       | Binomial          | Poisson            | Normal              |
|---------------|-------------------|--------------------|---------------------|
| Type          | Discrete          | Discrete           | Continuous          |
| Parameters    | n, p              | lambda             | mu, sigma           |
| Shape         | Skewed/Symmetric  | Right-skewed       | Perfectly symmetric |
| Example       | Coin flips        | Arrivals per hour  | Human heights       |

### 2. Real-World Significance of the Poisson Distribution

The Poisson distribution is critically useful whenever we count **rare or independent events over a fixed period or region**.

* **Telecommunications**: Number of calls arriving at a call center per minute.
* **Healthcare**: Number of patients arriving at an emergency room per hour.
* **Traffic Engineering**: Number of vehicles passing a checkpoint per minute.
* **Web Analytics**: Number of website hits per second.
* **Manufacturing**: Number of defects found in a batch of products.
* **Natural Events**: Number of earthquakes in a region per year.

**Why it matters**: It allows analysts and engineers to predict probabilities of rare events and plan capacity (staff, servers, inventory) accordingly.

### 3. Central Limit Theorem (CLT) and Its Relation to Normal Distribution

* **Central Limit Theorem (CLT)**:
    * States that if you take sufficiently large random samples (n >= 30) from **any population** (regardless of its original distribution), the distribution of the **sample means** will approximate a **Normal distribution**.
    * The mean of sample means equals the population mean.
    * The standard deviation of sample means (standard error) = sigma / sqrt(n).

* **Relation to Normal Distribution**:
    * CLT **justifies** the widespread use of the Normal distribution in statistics.
    * Even if the underlying data is Binomial, Poisson, or Uniform, with enough samples, the sampling distribution becomes Normal.
    * This is why Normal distribution is the backbone of hypothesis testing, confidence intervals, and regression.

* **Practical implication**: You can apply Normal-based statistical tools to almost any data if your sample size is large enough.

### 4. Why is Normal Distribution Widely Used in Machine Learning and Statistics?

* **Mathematical Convenience**: Many statistical operations (mean, variance, linear combinations) have closed-form solutions under normality.
* **Central Limit Theorem**: Aggregated data tends to be normally distributed regardless of source distribution.
* **Maximum Entropy**: Given a fixed mean and variance, the Normal distribution assumes the least additional information — the most natural choice.
* **Gaussian Noise Model**: Real-world measurement errors and noise are commonly assumed to be normally distributed.
* **Algorithms Built on Normality**:
    * **Linear Regression**: Assumes residuals are normally distributed.
    * **Naive Bayes**: Uses Gaussian distribution for continuous features.
    * **PCA, LDA**: Based on Gaussian assumptions.
    * **Neural Networks**: Weight initialization often uses Normal distribution.
* **Parameter Estimation**: MLE (Maximum Likelihood Estimation) under normality leads to simple closed-form solutions.

### 5. Skewness in Distributions

* **Definition**: Skewness measures the **asymmetry** of a probability distribution around its mean.
    * **Positive Skew (Right-skewed)**: Tail is longer on the right; mean > median > mode.
    * **Negative Skew (Left-skewed)**: Tail is longer on the left; mean < median < mode.
    * **Zero Skew**: Perfectly symmetric distribution.

* **Which distributions exhibit skewness?**
    * **Poisson Distribution**: Always right-skewed (positive skew), more so when lambda is small.
    * **Binomial Distribution**: Right-skewed when p < 0.5, left-skewed when p > 0.5, symmetric when p = 0.5.
    * **Normal Distribution**: Zero skewness — perfectly symmetric.
    * **Exponential Distribution**: Strongly right-skewed.
    * **Chi-Squared Distribution**: Right-skewed, especially with fewer degrees of freedom.

* **Why it matters**: Skewness affects which statistical methods are appropriate. Highly skewed data may require transformation (log, square root) before applying Normal-assumption-based techniques.

### Demonstrations of Probability Distributions

#### 1. Binomial Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Parameters
n = 20       # number of trials
p = 0.5      # probability of success (fair coin)

# Generate Binomial distribution
x = np.arange(0, n + 1)
pmf = stats.binom.pmf(x, n, p)

# Plot
plt.figure(figsize=(8, 4))
plt.bar(x, pmf, color='steelblue', edgecolor='black', alpha=0.8)
plt.title(f'Binomial Distribution (n={n}, p={p})')
plt.xlabel('Number of Successes')
plt.ylabel('Probability')
plt.xticks(x)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# Statistics
mean, var = stats.binom.stats(n, p)
print(f"Binomial Distribution Statistics:")
print(f"  Mean     : {mean}")
print(f"  Variance : {var}")
print(f"  Std Dev  : {np.sqrt(var):.4f}")
print(f"  P(X = 10): {stats.binom.pmf(10, n, p):.4f}")
print(f"  P(X <= 8): {stats.binom.cdf(8, n, p):.4f}")
```

#### 2. Poisson Distribution

```python
# Poisson Distribution
# Real-world scenario: Average 5 customers arrive per hour

lambda_val = 5   # average rate (lambda)
x = np.arange(0, 20)
pmf = stats.poisson.pmf(x, lambda_val)

# Plot
plt.figure(figsize=(8, 4))
plt.bar(x, pmf, color='coral', edgecolor='black', alpha=0.8)
plt.title(f'Poisson Distribution (lambda = {lambda_val})')
plt.xlabel('Number of Events')
plt.ylabel('Probability')
plt.xticks(x)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# Statistics
mean, var = stats.poisson.stats(lambda_val)
print(f"Poisson Distribution Statistics (lambda = {lambda_val}):")
print(f"  Mean     : {mean}")
print(f"  Variance : {var}")
print(f"  Std Dev  : {np.sqrt(var):.4f}")
print(f"  P(X = 5) : {stats.poisson.pmf(5, lambda_val):.4f}")
print(f"  P(X <= 3): {stats.poisson.cdf(3, lambda_val):.4f}")

# Compare different lambda values
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, lam in zip(axes, [2, 5, 10]):
    pmf_vals = stats.poisson.pmf(np.arange(0, 25), lam)
    ax.bar(np.arange(0, 25), pmf_vals, color='coral', edgecolor='black', alpha=0.8)
    ax.set_title(f'Poisson (lambda = {lam})')
    ax.set_xlabel('Events')
    ax.set_ylabel('Probability')
    ax.grid(axis='y', linestyle='--', alpha=0.6)
plt.suptitle('Poisson Distribution with Different Lambda Values')
plt.tight_layout()
plt.show()
```

#### 3. Normal Distribution

```python
# Normal Distribution
# Real-world scenario: Heights of adult men (mean=170cm, std=10cm)

mu = 170     # mean
sigma = 10   # standard deviation
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 300)
pdf = stats.norm.pdf(x, mu, sigma)

# Plot
plt.figure(figsize=(9, 4))
plt.plot(x, pdf, color='green', linewidth=2.5)
plt.fill_between(x, pdf, alpha=0.15, color='green')

# Shade 1, 2, 3 sigma regions
for k, color, label in [(1, 'green', '68.27% (1 sigma)'),
                         (2, 'blue',  '95.45% (2 sigma)'),
                         (3, 'red',   '99.73% (3 sigma)')]:
    x_fill = np.linspace(mu - k*sigma, mu + k*sigma, 300)
    plt.fill_between(x_fill, stats.norm.pdf(x_fill, mu, sigma),
                     alpha=0.15, color=color, label=label)

plt.axvline(mu, color='black', linestyle='--', linewidth=1.5, label=f'Mean = {mu}')
plt.title(f'Normal Distribution (mu = {mu}, sigma = {sigma})')
plt.xlabel('Height (cm)')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# Statistics
print(f"Normal Distribution Statistics (mu={mu}, sigma={sigma}):")
print(f"  P(X < 180)        : {stats.norm.cdf(180, mu, sigma):.4f}")
print(f"  P(160 < X < 180)  : {stats.norm.cdf(180, mu, sigma) - stats.norm.cdf(160, mu, sigma):.4f}")
print(f"  P(X > 190)        : {1 - stats.norm.cdf(190, mu, sigma):.4f}")
print(f"  95th percentile   : {stats.norm.ppf(0.95, mu, sigma):.2f} cm")
```

#### 4. Demonstrating the Central Limit Theorem

```python
# Central Limit Theorem Demonstration
# Starting from a skewed Exponential distribution

np.random.seed(42)

# Original population: Exponential (right-skewed)
population = np.random.exponential(scale=2, size=100000)

sample_sizes = [1, 5, 30, 100]
num_samples = 2000

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, n_size in zip(axes, sample_sizes):
    sample_means = [np.mean(np.random.choice(population, size=n_size))
                    for _ in range(num_samples)]
    ax.hist(sample_means, bins=40, color='mediumpurple', edgecolor='black', alpha=0.8, density=True)
    ax.set_title(f'Sample Size n = {n_size}')
    ax.set_xlabel('Sample Mean')
    ax.set_ylabel('Density')
    ax.grid(linestyle='--', alpha=0.5)

plt.suptitle('Central Limit Theorem: Exponential Population → Normal Sample Means', fontsize=12)
plt.tight_layout()
plt.show()

print("As sample size increases, the distribution of sample means becomes Normal.")
print("This holds regardless of the original population distribution (CLT).")
```

#### 5. Comparing Skewness Across Distributions

```python
from scipy.stats import skew

np.random.seed(0)

# Generate data from each distribution
binomial_data = np.random.binomial(n=20, p=0.3, size=5000)   # right-skewed (p < 0.5)
poisson_data  = np.random.poisson(lam=2, size=5000)           # right-skewed
normal_data   = np.random.normal(loc=0, scale=1, size=5000)   # symmetric

datasets = [
    (binomial_data, 'Binomial (n=20, p=0.3)', 'steelblue'),
    (poisson_data,  'Poisson (lambda=2)',      'coral'),
    (normal_data,   'Normal (mu=0, sigma=1)',   'green'),
]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, (data, title, color) in zip(axes, datasets):
    ax.hist(data, bins=30, color=color, edgecolor='black', alpha=0.8, density=True)
    ax.set_title(f'{title}\nSkewness = {skew(data):.3f}')
    ax.set_xlabel('Value')
    ax.set_ylabel('Density')
    ax.grid(linestyle='--', alpha=0.5)

plt.suptitle('Skewness Comparison: Binomial vs Poisson vs Normal', fontsize=13)
plt.tight_layout()
plt.show()

print("Skewness Summary:")
for data, title, _ in datasets:
    s = skew(data)
    direction = "Right-skewed" if s > 0.1 else ("Left-skewed" if s < -0.1 else "Symmetric")
    print(f"  {title}: Skewness = {s:.4f} → {direction}")
```
